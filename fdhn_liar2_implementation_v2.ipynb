# LIAR2 FDHN regularization experiment: weight decay 5e-4

Runs the corrected LIAR2 model with AdamW weight decay set to `5e-4`, validation-accuracy checkpoint selection, and early stopping after two epochs without validation-accuracy improvement. This changes only weight decay relative to the prior `1e-4` regularized run.

The recorded paper-reference result is **70.21% test accuracy** (macro-F1 **69.61%**). This run is an experiment, not a guaranteed improvement. Its checkpoint is saved to `liar2/LIAR2_wd5e-4.pt`, leaving both earlier checkpoints untouched.

The script requires PyTorch, scikit-learn, `transformers`, and access to the `bert-base-uncased` tokenizer files (local cache or internet). The prior regularized CPU run took about 28 minutes; runtime depends on hardware.

Run the code cell below from any directory inside this repository.

```python
import os
import runpy
from pathlib import Path

script_name = Path("segmentation_experiments/5.1.LIAR2.py")
repo_root = next(
    (candidate for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (candidate / script_name).is_file()),
    None,
)
if repo_root is None:
    raise FileNotFoundError(f"Could not locate {script_name} from {Path.cwd()}")

data_dir = repo_root / "liar2"
missing_splits = [name for name in ("train.csv", "valid.csv", "test.csv")
                  if not (data_dir / name).is_file()]
if missing_splits:
    raise FileNotFoundError(f"Missing LIAR2 split(s) in {data_dir}: {missing_splits}")

previous_cwd = Path.cwd()
managed_environment = {
    "LIAR2_REGULARIZED_TRAINING": "1",
    "LIAR2_WEIGHT_DECAY": "5e-4",
    "LIAR2_REGULARIZED_TAG": "wd5e-4",
}
previous_environment = {key: os.environ.get(key) for key in managed_environment}
try:
    os.chdir(data_dir)
    os.environ.update(managed_environment)
    runpy.run_path(str(repo_root / script_name), run_name="__main__")
finally:
    os.chdir(previous_cwd)
    for key, value in previous_environment.items():
        if value is None:
            os.environ.pop(key, None)
        else:
            os.environ[key] = value
```
