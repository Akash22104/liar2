```python
import pandas as pd

DATA_DIR = "liar2"

train_data = pd.read_csv(f"{DATA_DIR}/train.csv")
valid_data = pd.read_csv(f"{DATA_DIR}/valid.csv")
test_data = pd.read_csv(f"{DATA_DIR}/test.csv")
```

```python
from IPython.display import display

# Display the first rows and the column names.
display(train_data.head())
display(pd.DataFrame({"column": train_data.columns}))
```

```python
display(pd.DataFrame({
    "split": ["train", "validation", "test"],
    "rows": [train_data.shape[0], valid_data.shape[0], test_data.shape[0]],
    "columns": [train_data.shape[1], valid_data.shape[1], test_data.shape[1]],
}))
```

```python
print("Train shape:", train_data.shape)
print("Validation shape:", valid_data.shape)
print("Test shape:", test_data.shape)
```

```python
print("Train info:")
train_data.info()

print("\nValidation info:")
valid_data.info()

print("\nTest info:")
test_data.info()

print("\nBasic statistics for numeric columns in the training set:")
display(train_data.describe())
```
